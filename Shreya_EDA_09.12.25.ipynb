```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

# Load datasets
day = pd.read_csv("day_mycopy.csv")
hour = pd.read_csv("hour_mycopy.csv")

print("DAY dataset shape:", day.shape)
print("HOUR dataset shape:", hour.shape)

print("\nDAY dataset info:")
print(day.info())

print("\nHOUR dataset info:")
print(hour.info())
```

```python
print("\nNull values in DAY:\n", day.isnull().sum())
print("\nNull values in HOUR:\n", hour.isnull().sum())

print("\nDuplicate rows in DAY:", day.duplicated().sum())
print("Duplicate rows in HOUR:", hour.duplicated().sum())

print("\nDescriptive statistics (DAY):")
display(day.describe())

print("\nDescriptive statistics (HOUR):")
display(hour.describe())

```

```python
print("DAILY RENTAL DISTRIBUTION (SUMMARY) ")

day_mean = day['cnt'].mean()
day_median = day['cnt'].median()
day_min = day['cnt'].min()
day_max = day['cnt'].max()
day_std = day['cnt'].std()
day_cv = day_std / day_mean   # coefficient of variation

print(f"Minimum daily rentals        : {day_min}")
print(f"Maximum daily rentals        : {day_max}")
print(f"Average (mean) daily rentals : {day_mean:.2f}")
print(f"Median daily rentals         : {day_median:.2f}")
print(f"Standard deviation           : {day_std:.2f}")
print(f"Coefficient of Variation (CV): {day_cv:.2f}")
print("Interpretation: CV shows how stable or unstable daily demand is.\n")


print(" HOURLY RENTAL DISTRIBUTION (SUMMARY) ")

hour_mean = hour['cnt'].mean()
hour_median = hour['cnt'].median()
hour_min = hour['cnt'].min()
hour_max = hour['cnt'].max()
hour_std = hour['cnt'].std()
hour_cv = hour_std / hour_mean

print(f"Minimum hourly rentals        : {hour_min}")
print(f"Maximum hourly rentals        : {hour_max}")
print(f"Average (mean) hourly rentals : {hour_mean:.2f}")
print(f"Median hourly rentals         : {hour_median:.2f}")
print(f"Standard deviation            : {hour_std:.2f}")
print(f"Coefficient of Variation (CV) : {hour_cv:.2f}\n")


print(" Insights ")
print("- Daily values capture long-term stable demand patterns.")
print("- Hourly values show big fluctuations because usage depends on time of day.")
print("- Higher CV in hourly data → demand is much less predictable hour-to-hour.")
print("- Lower CV in daily data → daily demand is more consistent and aggregated.")
```

```python
avg_season = day.groupby("season")["cnt"].mean()

plt.figure(figsize=(7,4))
sns.barplot(x=avg_season.index, y=avg_season.values)
plt.title("Average Rentals by Season")
plt.xlabel("Season")
plt.ylabel("Average Count")
plt.show()

print("\nINSIGHT:")
print("Highest season:", avg_season.idxmax())
print("Lowest season:", avg_season.idxmin())
```

```python
avg_weather = day.groupby("weathersit")["cnt"].mean()

plt.figure(figsize=(7,4))
sns.barplot(x=avg_weather.index, y=avg_weather.values)
plt.title("Average Rentals by Weather Condition")
plt.xlabel("Weather Code")
plt.ylabel("Average Count")
plt.show()

print("\nINSIGHT:")
print("Best weather:", avg_weather.idxmax())
print("Worst weather:", avg_weather.idxmin())

```

```python
avg_working = day.groupby("workingday")["cnt"].mean()

plt.figure(figsize=(6,4))
sns.barplot(x=avg_working.index, y=avg_working.values)
plt.xticks([0,1], ["Weekend/Holiday", "Working Day"])
plt.title("Working Day vs Weekend Rentals")
plt.ylabel("Average Count")
plt.show()

print("\nInsight:")
if avg_working[1] > avg_working[0]:
    print("People ride more on working days (commuting effect).")
else:
    print("People ride more on weekends (leisure effect).")
```

```python
avg_hour = hour.groupby("hr")["cnt"].mean()

plt.figure(figsize=(10,5))
sns.lineplot(x=avg_hour.index, y=avg_hour.values, marker="o")
plt.title("Average Rentals by Hour of Day")
plt.xlabel("Hour")
plt.ylabel("Average Count")
plt.grid(True)
plt.show()

print("\nPEAK HOUR:", avg_hour.idxmax())
print("LOWEST HOUR:", avg_hour.idxmin())
```

```python
avg_year = day.groupby("yr")["cnt"].mean()

plt.figure(figsize=(6,4))
sns.barplot(x=avg_year.index, y=avg_year.values)
plt.xticks([0,1], ["2011", "2012"])
plt.title("Yearly Average Rentals")
plt.ylabel("Average Count")
plt.show()

print("\nINSIGHT:")
print("2011:", avg_year[0])
print("2012:", avg_year[1])
print("Growth %:", round(((avg_year[1] - avg_year[0]) / avg_year[0]) * 100, 2))
```

```python
plt.figure(figsize=(7,4))
sns.scatterplot(x=day["temp"], y=day["cnt"], alpha=0.4)
sns.regplot(x=day["temp"], y=day["cnt"], scatter=False, color="red")
plt.title("Temperature vs Rentals")
plt.show()

print("\nINSIGHT:")
print("Demand increases with temperature until it becomes too hot.")
```
