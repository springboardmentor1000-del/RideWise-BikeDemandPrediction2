```python
plt.figure(figsize=(6,4))
plt.hist(['cnt'])
plt.title("Distribution of Total Daily Rentals")
plt.xlabel("Total Rentals (cnt)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(6,4))
plt.hist(day['temp'])
plt.title("Distribution of Temperature")
plt.xlabel("Temperature")
plt.ylabel("Frequency")
#plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(day['hum'])
plt.title("Distribution of Humidity")
plt.xlabel("Humidity")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))
day.groupby('season')['cnt'].mean().plot(kind='bar')
plt.title("Average Rentals by Season")
plt.xlabel("Season")
plt.ylabel("Average Rentals")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))
day.groupby('weathersit')['cnt'].mean().plot(kind='bar')
plt.title("Average Rentals by Weather Situation")
plt.xlabel("Weather Category")
plt.ylabel("Average Rentals")
plt.tight_layout()
plt.show()
```

**Bivariate** **analysis**

```python
plt.figure(figsize=(6,4))
plt.scatter(day['temp'], day['cnt'])
plt.title("Temperature vs Total Rentals")
plt.xlabel("Temperature")
plt.ylabel("Total Rentals")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(day['hum'], day['cnt'])
plt.title("Humidity vs Total Rentals")
plt.xlabel("Humidity")
plt.ylabel("Total Rentals")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(day['windspeed'], day['cnt'])
plt.title("Wind Speed vs Total Rentals")
plt.xlabel("Wind Speed")
plt.ylabel("Total Rentals")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,4))
day.groupby('mnth')['cnt'].mean().plot(kind='line', marker='o')
plt.title("Monthly Trend of Bike Rentals")
plt.xlabel("Month")
plt.ylabel("Average Rentals")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))
day.groupby('yr')['cnt'].sum().plot(kind='bar')
plt.title("Yearly Total Rentals")
plt.xlabel("Year (0=2011, 1=2012)")
plt.ylabel("Total Rentals")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))
day.groupby('workingday')['cnt'].mean().plot(kind='bar')
plt.title("Working Day vs Weekend Rentals")
plt.xlabel("Working Day (1=Yes)")
plt.ylabel("Average Rentals")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.bar(day['casual'], day['registered'], alpha=0.6)
plt.title("Casual vs Registered Users")
plt.xlabel("Casual Users")
plt.ylabel("Registered Users")
plt.grid(True)
plt.show()
```

```python
corr = day.corr(numeric_only=True)

plt.figure(figsize=(10,8))
plt.imshow(corr, interpolation='nearest')
plt.title("Correlation Heatmap")
plt.colorbar()
plt.xticks(np.arange(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(np.arange(len(corr.columns)), corr.columns)
plt.tight_layout()
plt.show()
```
