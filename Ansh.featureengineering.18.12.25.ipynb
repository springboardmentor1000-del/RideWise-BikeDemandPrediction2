```python
import pandas as pd
import os

# Define the path to the hour.csv file
hour_csv_file_path = 'Dataset exploration/hour.csv'

# Check if the file exists
if os.path.exists(hour_csv_file_path):
    df_hour = pd.read_csv(hour_csv_file_path, encoding='unicode_escape')
    print("Successfully loaded hour.csv:")
    display(df_hour.head())
else:
    print(f"Error: '{hour_csv_file_path}' not found. Please ensure the file is extracted.")
```

```python
# convert date
df['dteday'] = pd.to_datetime(df['dteday'])
df_hour['dteday'] = pd.to_datetime(df_hour['dteday'])

for dframe in [df, df_hour]:
    dframe['day'] = dframe['dteday'].dt.day
    dframe['month'] = dframe['dteday'].dt.month
    dframe['year'] = dframe['dteday'].dt.year
    dframe['dayofweek'] = dframe['dteday'].dt.dayofweek
    dframe['is_weekend'] = dframe['dayofweek'].isin([5,6]).astype(int)
```

```python
import numpy as np

for df_current in [df, df_hour]:
    df_current['month_sin'] = np.sin(2 * np.pi * df_current['month']/12)
    df_current['month_cos'] = np.cos(2 * np.pi * df_current['month']/12)
```

```python
for df_current in [df, df_hour]:
    df_current['weather_severity'] = (
        df_current['weathersit'] *
        (1 + df_current['windspeed']) *
        (1 + df_current['hum'])
    )
```

```python
df.head()
```

```python
for df_current in [df, df_hour]:
    df_current['comfort_index'] = (
        (1 - abs(df_current['temp'] - df_current['atemp'])) *
        (1 - df_current['hum']) *
        (1 - df_current['windspeed'])
    )
```

```python
for df_current in [df, df_hour]:
    df_current['is_peak_season'] = df_current['month'].isin([5, 6, 7, 8]).astype(int)
```

```python
# Define rush hours (e.g., 7-9 AM and 5-7 PM)
df_hour['is_rush_hour'] = ((df_hour['hr'] >= 7) & (df_hour['hr'] <= 9)) | ((df_hour['hr'] >= 17) & (df_hour['hr'] <= 19)).astype(int)
```
