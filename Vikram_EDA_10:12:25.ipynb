```python
!pip install -q plotly
import plotly.express as px
import pandas as pd
df = df_hour.copy()
season_map = {1:"Spring", 2:"Summer", 3:"Fall", 4:"Winter"}
weather_map = {1:"Clear", 2:"Mist/Cloudy", 3:"LightRain/Snow", 4:"HeavyRain/Snow"}
df['season_name'] = df['season'].map(season_map)
df['weather_name'] = df['weathersit'].map(weather_map)
df['is_weekend'] = df['weekday'].apply(lambda x: "Weekend" if x in [0,6] else "Weekday")
df['working_status'] = df['workingday'].map({1:"Working Day", 0:"Holiday/Weekend"})
df['month'] = pd.to_datetime(df['dteday']).dt.month
agg = df.groupby(['season_name','month','weather_name','working_status','is_weekend'], as_index=False) \
        .agg(total_rentals=('cnt','sum'), avg_rentals=('cnt','mean'))
fig = px.treemap(
    agg,
    path=['season_name','month','weather_name','working_status','is_weekend'],
    values='total_rentals',
    color='avg_rentals',
    color_continuous_scale='YlOrRd',
    title='map: Season → Month → Weather → Working Status → Weekend (Size=total rentals, Color=avg rentals)'
)
fig.update_traces(textinfo="label+value")
fig.show()

```

```python
df_day = df_day.drop(columns=["casual", "registered"])
```

```python
df_day["dteday"] = pd.to_datetime(df_day["dteday"])

df_day["year"] = df_day["dteday"].dt.year
df_day["month"] = df_day["dteday"].dt.month
df_day["day"] = df_day["dteday"].dt.day
df_day["weekday"] = df_day["dteday"].dt.weekday
df_day["is_weekend"] = df_day["weekday"].isin([5, 6]).astype(int)
```

```python
season_map = {1: "spring", 2: "summer", 3: "fall", 4: "winter"}
weather_map = {1: "clear", 2: "mist", 3: "light_rain", 4: "heavy_rain"}

df_day["season_label"] = df_day["season"].map(season_map)
df_day["weather_label"] = df_day["weathersit"].map(weather_map)
```

```python
df_day["temp_humidity"] = df_day["temp"] * df_day["hum"]
df_day["temp_windspeed"] = df_day["temp"] * df_day["windspeed"]
```

```python
df_day["is_working_day"] = ((df_day["workingday"] == 1) & (df_day["holiday"] == 0)).astype(int)
df_day["holiday_or_weekend"] = ((df_day["holiday"] == 1) | (df_day["is_weekend"] == 1)).astype(int)
```

```python
df_day = df_day.sort_values("dteday")

df_day["cnt_lag_1"] = df_day["cnt"].shift(1)
df_day["cnt_lag_7"] = df_day["cnt"].shift(7)
```

```python
df_day["rolling_mean_7"] = df_day["cnt"].rolling(window=7).mean()
```

```python
df_day = pd.get_dummies(
    df_day,
    columns=["season_label", "weather_label"],
    drop_first=True
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scale_cols = ["temp", "hum", "windspeed", "temp_humidity", "temp_windspeed"]
df_day[scale_cols] = scaler.fit_transform(df_day[scale_cols])
```

```python
df_hour = df_hour.drop(columns=["casual", "registered"])
```

```python
df_hour["dteday"] = pd.to_datetime(df_hour["dteday"])

df_hour["year"] = df_hour["dteday"].dt.year
df_hour["month"] = df_hour["dteday"].dt.month
df_hour["day"] = df_hour["dteday"].dt.day
df_hour["weekday"] = df_hour["dteday"].dt.weekday
df_hour["is_weekend"] = df_hour["weekday"].isin([5, 6]).astype(int)

df_hour["hour"] = df_hour["hr"]
```

```python
df_hour["is_peak_hour"] = df_hour["hour"].isin([7, 8, 9, 17, 18, 19]).astype(int)
df_hour["is_morning"] = df_hour["hour"].isin([6, 7, 8, 9, 10, 11]).astype(int)
df_hour["is_evening"] = df_hour["hour"].isin([16, 17, 18, 19, 20]).astype(int)
df_hour["is_night"] = df_hour["hour"].isin([0, 1, 2, 3, 4, 5]).astype(int)
```

```python
season_map = {1: "spring", 2: "summer", 3: "fall", 4: "winter"}
weather_map = {1: "clear", 2: "mist", 3: "light_rain", 4: "heavy_rain"}

df_hour["season_label"] = df_hour["season"].map(season_map)
df_hour["weather_label"] = df_hour["weathersit"].map(weather_map)
```

```python
df_hour["temp_humidity"] = df_hour["temp"] * df_hour["hum"]
df_hour["temp_windspeed"] = df_hour["temp"] * df_hour["windspeed"]
```

```python
df_hour["is_working_day"] = ((df_hour["workingday"] == 1) & (df_hour["holiday"] == 0)).astype(int)
df_hour["holiday_or_weekend"] = ((df_hour["holiday"] == 1) | (df_hour["is_weekend"] == 1)).astype(int)
```

```python
df_hour = df_hour.sort_values(["dteday", "hour"])

df_hour["cnt_lag_1"] = df_hour["cnt"].shift(1)
df_hour["cnt_lag_7"] = df_hour["cnt"].shift(24 * 7)
```

```python
df_hour["rolling_mean_3"] = df_hour["cnt"].rolling(window=3).mean()
df_hour["rolling_mean_24"] = df_hour["cnt"].rolling(window=24).mean()
```

```python
df_hour = pd.get_dummies(
    df_hour,
    columns=["season_label", "weather_label"],
    drop_first=True
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scale_cols = ["temp", "hum", "windspeed", "temp_humidity", "temp_windspeed"]
df_hour[scale_cols] = scaler.fit_transform(df_hour[scale_cols])
```

```python
df_hour = df_hour.drop(columns=["instant", "dteday", "hr"])
df_hour = df_hour.dropna()
```
