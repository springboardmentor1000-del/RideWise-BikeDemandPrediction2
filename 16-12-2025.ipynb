```python
import pandas as pd
import numpy as np
day_df = pd.read_csv("day.csv")
```

```python
# Helps etract meaningful date-based features.
# Models cannot understand raw date strings.
# Datetime allows extraction of seasonal and trend-based features.
day_df['dteday'] = pd.to_datetime(day_df['dteday'])
# Helps extract meaningful date-based features.
```

```python
# Captures seasonal & weekly patterns.
# Bike rentals depend on day, month, year, and weekday patterns.
# learn seasonal behavior.
day_df['day'] = day_df['dteday'].dt.day
day_df['month'] = day_df['dteday'].dt.month
day_df['year'] = day_df['dteday'].dt.year
day_df['weekday'] = day_df['dteday'].dt.weekday
```

```python
# Shows usage behavior difference between casual and registered users.
# Understanding the ratio between casual and registered users gives 
# better demand insight than raw counts.
day_df['casual_registered_ratio'] = day_df['casual'] / (day_df['registered'] + 1)
```

```python
# Removes identifiers that don’t help prediction.
# instant is just an ID
# dteday is already converted into useful feature
day_df.drop(columns=['instant', 'dteday'], inplace=True)
```

```python
import pandas as pd

hour_df = pd.read_csv("hour.csv")
```

```python
# Convert date column
# Hourly data still follows daily & seasonal trends.
hour_df['dteday'] = pd.to_datetime(hour_df['dteday'])
```

```python
# Create date-based features
# To capture weekly and monthly patterns affecting hourly demand
hour_df['day'] = hour_df['dteday'].dt.day
hour_df['month'] = hour_df['dteday'].dt.month
hour_df['year'] = hour_df['dteday'].dt.year
hour_df['weekday'] = hour_df['dteday'].dt.weekday
```

```python
# Create peak hour feature
# # Identifies rush hours (office time impact).
# Bike rentals increase during office hours and commute times.
# Binary features help models detect such patterns easily.
hour_df['is_peak_hour'] = hour_df['hr'].apply(
    lambda x: 1 if x in [7, 8, 9, 17, 18, 19] else 0
)
```

```python
# Drop unnecessary columns
# instant is an index
# dteday is replaced by extracted features
hour_df.drop(columns=['instant', 'dteday'], inplace=True)
```

```python
hour_df['hr_sin'] = np.sin(2 * np.pi * hour_df['hr'] / 24)
hour_df['hr_cos'] = np.cos(2 * np.pi * hour_df['hr'] / 24)
```
